## Project Overview

We will build a library management system. The system will allow users to:
- Add books to the library
- Borrow books
- Return books
- Search for books
- Display all books in the library

## Project Requirements

1. Create a `Book` class with the following attributes:
   - `title`
   - `author`
   - `year_published`
   - `genre`
   - `status` (available or borrowed)
2. Create a `Library` class with the following attributes:
   - `books` (a list of `Book` objects)

```python
# Case Study: Library Management System

# Singleton for Database Connection
class DatabaseConnection:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            print("Database connected.")
        return cls._instance

# Strategy for Payment Methods
class PaymentStrategy:
    def pay(self, amount): pass

class CashPayment(PaymentStrategy):
    def pay(self, amount):
        return f"Paid {amount} in cash."

class CardPayment(PaymentStrategy):
    def pay(self, amount):
        return f"Paid {amount} by card."

# Observer for Notifications
class Observer:
    def update(self, message): pass

class Member(Observer):
    def __init__(self, name):
        self.name = name
    def update(self, message):
        print(f"{self.name} notified: {message}")

class Library:
    def __init__(self):
        self.members = []
        self.books = ["Python Basics", "OOP in Depth"]

    def attach(self, member):
        self.members.append(member)

    def notify(self, message):
        for m in self.members:
            m.update(message)

    def borrow_book(self, book, payment: PaymentStrategy):
        if book in self.books:
            self.books.remove(book)
            self.notify(f"Book borrowed: {book}")
            print(payment.pay(100))
        else:
            print("Book not available.")

# Usage
db = DatabaseConnection()

library = Library()
member1 = Member("Alice")
member2 = Member("Bob")

library.attach(member1)
library.attach(member2)

library.borrow_book("Python Basics", CashPayment())
library.borrow_book("OOP in Depth", CardPayment())
```

## Key Takeaways
- A **Library Management System** can combine multiple OOP concepts:
  - Singleton for database connection
  - Strategy for flexible payment methods
  - Observer for notifying members
- This case study shows how design patterns work together in a real-world scenario.

## Code Explanation

- **Strategy Pattern**: Allows for flexible payment methods.
