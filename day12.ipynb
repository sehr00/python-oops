# Day 12: Operator Overloading in Python

In this notebook, we will learn:
- What operator overloading means in Object‑Oriented Programming.
- How to redefine operators like `+`, `-`, `*` for custom classes.
- Why operator overloading makes objects behave more naturally.


```python
# Example: Operator Overloading

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overload the + operator
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    # Overload the - operator
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    # Overload the * operator
    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# Usage
v1 = Vector(2, 3)
v2 = Vector(4, 5)

print(v1 + v2)   # Vector(6, 8)
print(v2 - v1)   # Vector(2, 2)
print(v1 * 3)    # Vector(6, 9)
```

## Key Takeaways
- Operator overloading lets you redefine how operators work for your classes.
- In the example, `+`, `-`, and `*` are overloaded for the `Vector` class.
- This makes objects behave more naturally and improves readability.
