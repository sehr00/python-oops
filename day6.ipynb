# Day 6: Encapsulation in Python

In this notebook, we will learn:
- What encapsulation means in Object‑Oriented Programming.
- How to use public, protected, and private attributes in Python.
- Why encapsulation helps keep data safe and controlled.


```python
# Example: Encapsulation

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # public attribute
        self._balance = balance     # protected attribute
        self.__pin = "1234"         # private attribute

    def deposit(self, amount):
        self._balance += amount
        print(f"Deposited {amount}. New balance: {self._balance}")

    def withdraw(self, amount, pin):
        if pin == self.__pin:
            if amount <= self._balance:
                self._balance -= amount
                print(f"Withdrew {amount}. Remaining balance: {self._balance}")
            else:
                print("Insufficient funds!")
        else:
            print("Invalid PIN!")

# Usage
account = BankAccount("S", 1000)
account.deposit(500)

# Accessing public attribute
print(account.owner)

# Accessing protected attribute (possible but discouraged)
print(account._balance)

# Accessing private attribute (will cause error)
# print(account.__pin)   # AttributeError
```

## Key Takeaways
- Public attributes can be accessed freely.
- Protected attributes (with `_`) are meant for internal use.
- Private attributes (with `__`) are hidden from outside the class.
- Encapsulation ensures data security and controlled access.
