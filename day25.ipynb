## What is Decorator Pattern?

The Decorator design pattern is a structural design pattern that allows you to add new behaviors to an object without altering its structure. It is used when you want to add functionality to an object dynamically, without modifying the object's code.

## Why use Decorator Pattern?

The Decorator pattern is used when you want to add new behaviors to an object dynamically, without modifying the object's code. It is useful when you want to add functionality to an object that is not always needed, or when you want to add functionality that can be easily removed.
## How to implement Decorator in Python?

To implement the Decorator pattern in Python, you need to create a decorator class that wraps the original object and adds new behavior. The decorator class should have an __init__ method that takes the original object as an argument, and a __call__ method that adds the new behavior.


```python
# Decorator Pattern Example

from abc import ABC, abstractmethod

# Component Interface
class Coffee(ABC):
    @abstractmethod
    def cost(self):
        pass

# Concrete Component
class SimpleCoffee(Coffee):
    def cost(self):
        return 50

# Decorator Base
class CoffeeDecorator(Coffee):
    def __init__(self, coffee: Coffee):
        self._coffee = coffee

    def cost(self):
        return self._coffee.cost()

# Concrete Decorators
class MilkDecorator(CoffeeDecorator):
    def cost(self):
        return self._coffee.cost() + 20

class SugarDecorator(CoffeeDecorator):
    def cost(self):
        return self._coffee.cost() + 10

class WhipDecorator(CoffeeDecorator):
    def cost(self):
        return self._coffee.cost() + 30

# Usage
coffee = SimpleCoffee()
print("Simple Coffee:", coffee.cost())

coffee_with_milk = MilkDecorator(coffee)
print("Coffee + Milk:", coffee_with_milk.cost())

coffee_with_milk_sugar = SugarDecorator(coffee_with_milk)
print("Coffee + Milk + Sugar:", coffee_with_milk_sugar.cost())

fancy_coffee = WhipDecorator(coffee_with_milk_sugar)
print("Coffee + Milk + Sugar + Whip:", fancy_coffee.cost())
```

## Key Takeaways
- The **Decorator pattern** allows behavior to be added to objects dynamically.
- It promotes flexibility by avoiding subclass explosion.
- Useful when you want to extend functionality without modifying existing code.
- Example: Adding milk, sugar, and whip cream to coffee step by step.
- Concrete decorators implement the additional behavior.
