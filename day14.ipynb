# Day 14: Properties and Getters/Setters

In this notebook, we will learn:
- How to use `@property` to create managed attributes.
- How getters and setters work in Python.
- Why properties are better than direct attribute access.


```python
# Example: Properties and Getters/Setters

class Employee:
    def __init__(self, name, salary):
        self._name = name
        self._salary = salary   # private attribute

    # Getter
    @property
    def salary(self):
        return self._salary

    # Setter
    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be negative!")
        self._salary = value

    # Deleter
    @salary.deleter
    def salary(self):
        print("Deleting salary...")
        del self._salary

# Usage
emp = Employee("Rahil", 50000)

print(emp.salary)       # Calls getter
emp.salary = 60000      # Calls setter
print(emp.salary)

# Uncomment to test deleter
# del emp.salary
```

## Key Takeaways
- Use `@property` to define getters in Python.
- Use `@<property>.setter` to define setters for controlled updates.
- Use `@<property>.deleter` to define deleters.
- Properties make code cleaner and safer than manually calling getter/setter methods.
