# Day 11: Magic/Dunder Methods in Python

In this notebook, we will learn:
- What magic/dunder methods are in Python.
- How they allow classes to behave like built‑in types.
- Common examples like `__str__`, `__repr__`, and `__add__`.


```python
# Example: Magic/Dunder Methods

class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    # __str__ gives a user-friendly string
    def __str__(self):
        return f"Book: {self.title}, Pages: {self.pages}"

    # __repr__ gives a developer-friendly string
    def __repr__(self):
        return f"Book(title='{self.title}', pages={self.pages})"

    # __add__ allows adding two Book objects
    def __add__(self, other):
        return self.pages + other.pages

# Usage
b1 = Book("Python Basics", 200)
b2 = Book("Advanced Python", 300)

print(str(b1))       # User-friendly output
print(repr(b1))      # Developer-friendly output
print(b1 + b2)       # Adds pages together
```

## Key Takeaways
- Magic methods start and end with double underscores (e.g., `__init__`, `__str__`).
- `__str__` is for readable output, `__repr__` is for debugging/developers.
- You can override operators like `+` using methods such as `__add__`.
- These methods make custom classes behave like built‑in Python objects.
