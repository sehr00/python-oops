# Day 3: Class vs. Instance Variables in Python

In this notebook, we will learn:
- The difference between **class variables** and **instance variables**.
- How class variables are shared across all objects.
- How instance variables are unique to each object.


```python
# Example: Class vs. Instance Variables

class Dog:
    # Class variable (shared by all instances)
    species = "Canis familiaris"

    def __init__(self, name, age):
        # Instance variables (unique to each object)
        self.name = name
        self.age = age

    def description(self):
        return f"{self.name} is {self.age} years old."

    def speak(self, sound):
        return f"{self.name} says {sound}"

# Creating objects
dog1 = Dog("Buddy", 3)
dog2 = Dog("Lucy", 5)

# Accessing class variable
print(dog1.species)   # Shared
print(dog2.species)   # Shared

# Accessing instance variables
print(dog1.description())
print(dog2.description())
```

## Key Concepts Learned
- **Class Variable**: Belongs to the class, shared by all objects (e.g., `species`).
- **Instance Variable**: Belongs to each object, unique per instance (e.g., `name`, `age`).
- **Accessing Variables**: Class variables can be accessed via the class or any object; instance variables are accessed via individual objects.
