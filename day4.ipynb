# Day 4: Inheritance in Python

In this notebook, we will learn:
- What inheritance is and why it’s useful.
- How a child class can inherit attributes and methods from a parent class.
- How to override methods in child classes.


```python
# Example: Single Inheritance

# Parent class
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound."

# Child class
class Dog(Animal):
    def speak(self):
        return f"{self.name} barks!"

class Cat(Animal):
    def speak(self):
        return f"{self.name} meows!"

# Creating objects
dog = Dog("Buddy")
cat = Cat("Lucy")

print(dog.speak())  # Buddy barks!
print(cat.speak())  # Lucy meows!
```

## Key Concepts Learned
- **Inheritance**: Mechanism where one class (child) derives from another (parent).
- **Parent Class**: The base class that provides attributes and methods.
- **Child Class**: The derived class that inherits and can override parent behavior.
- **Method Overriding**: Redefining a parent method in the child class.
