# Day 5: Multiple Inheritance and super()

In this notebook, we will learn:
- How a class can inherit from multiple parent classes.
- How Python resolves method calls using the **Method Resolution Order (MRO)**.
- How to use the `super()` function to call parent methods safely.


```python
# Example: Multiple Inheritance

class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"My name is {self.name}."

class Student:
    def __init__(self, student_id):
        self.student_id = student_id

    def study(self):
        return f"Student {self.student_id} is studying."

# Child class inherits from both Person and Student
class GraduateStudent(Person, Student):
    def __init__(self, name, student_id, degree):
        # Use super() to call parent constructors in MRO order
        Person.__init__(self, name)
        Student.__init__(self, student_id)
        self.degree = degree

    def details(self):
        return f"{self.introduce()} I am pursuing {self.degree}."

# Creating object
grad = GraduateStudent("Alice", "S123", "b.tech Computer Science")

print(grad.details())
print(grad.study())
```

## Key Concepts Learned
- **Multiple Inheritance**: A class can inherit from more than one parent class.
- **MRO (Method Resolution Order)**: Python decides which parent method to call first.
- **super()**: Used to safely call parent methods, especially in complex hierarchies.
