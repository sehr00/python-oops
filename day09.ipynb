# Day 9: Method Overriding in Python

In this notebook, we will learn:
- What method overriding means in Object‑Oriented Programming.
- How a child class can redefine a parent class method.
- Why overriding is useful for customizing behavior in subclasses.


```python
# Example: Method Overriding

class Animal:
    def speak(self):
        return "This animal makes a sound."

class Dog(Animal):
    def speak(self):
        return "Woof! Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

# Usage
animals = [Animal(), Dog(), Cat()]

for a in animals:
    print(a.speak())
```

## Key Takeaways
- Method overriding allows a child class to provide its own version of a method defined in the parent class.
- In the example, `speak()` is overridden in `Dog` and `Cat` to give specific sounds.
- Overriding makes code more flexible and allows subclasses to customize behavior.
