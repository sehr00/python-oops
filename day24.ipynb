## What is the Strategy Pattern?

The Strategy pattern is a behavioral design pattern that enables an object to alter its behavior at runtime by encapsulating alternative algorithms or strategies and making them interchangeable.

The Strategy pattern is useful when you have a class that performs multiple tasks, and you want to allow the user to choose which task to perform at runtime. This can be useful in situations where the behavior of a class can change based on the context in which it is used.

## Why use the Strategy Pattern?
The Strategy pattern is useful in situations where you have a class that performs multiple tasks, and you want to allow the user to choose which task to perform at runtime. This can be useful in situations where the behavior of a class can change based on the context in which it is used.

For example, consider a class that performs different types of sorting algorithms. You could use the Strategy pattern to allow the user to choose which sorting algorithm to use at runtime.

## How to implement the Strategy Pattern in Python?

To implement the Strategy pattern in Python, you can follow these steps:
1. Define a base class for the strategy, which will contain the common interface for all strategies.
2. Define a concrete strategy class for each alternative algorithm or strategy.
3. Define a context class that will use the strategy classes. The context class will have a reference to the strategy object and will call the strategy's method to perform the task.
4. Create instances of the strategy classes and pass them to the context class.
5. Call the method on the context class to perform the task using the chosen strategy.


```python
# Strategy Pattern Example

from abc import ABC, abstractmethod

# Strategy Interface
class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

# Concrete Strategies
class CreditCardPayment(PaymentStrategy):
    def pay(self, amount):
        return f"Paid {amount} using Credit Card"

class PayPalPayment(PaymentStrategy):
    def pay(self, amount):
        return f"Paid {amount} using PayPal"

class BankTransferPayment(PaymentStrategy):
    def pay(self, amount):
        return f"Paid {amount} using Bank Transfer"

# Context
class ShoppingCart:
    def __init__(self, strategy: PaymentStrategy):
        self.strategy = strategy

    def checkout(self, amount):
        return self.strategy.pay(amount)

# Usage
cart1 = ShoppingCart(CreditCardPayment())
cart2 = ShoppingCart(PayPalPayment())
cart3 = ShoppingCart(BankTransferPayment())

print(cart1.checkout(100))
print(cart2.checkout(200))
print(cart3.checkout(300))
```

## Key Takeaways
- The **Strategy pattern** defines a family of algorithms, encapsulates each one, and makes them interchangeable.
- The client (context) can switch strategies at runtime without changing its code.
- Useful when you want flexibility in choosing algorithms (e.g., payment methods, sorting techniques).
- Example: A shopping cart can use different payment strategies without modifying the checkout logic.
