## What is the Proxy Pattern?

The Proxy design pattern is a structural design pattern that allows you to provide a substitute or placeholder for another object. The proxy object controls access to the original object, providing a level of indirection between the client and the real object.

The Proxy pattern is useful when you want to add additional functionality to an existing object without modifying its code. It can be used to control access to a resource, such as a file or network connection, or to add logging or monitoring to an object.

## Why use the Proxy Pattern?
The Proxy pattern is useful in situations where you want to add additional functionality to an existing object without modifying its code. For example, you might want to add logging or monitoring to an object, or control access to a resource such as a file or network connection.

The Proxy pattern can also be used to provide a substitute for an object that is expensive to create or that is not available at the time of creation. For example, you might want to create a proxy object that loads a large image from a remote server only when it is needed.

## How to use the Proxy Pattern?
To use the Proxy pattern, you need to create a proxy object that implements the same interface as the original object. The proxy object should have a reference to the original object and should delegate method calls to the original object.




```python
# Proxy Pattern Example

from abc import ABC, abstractmethod

# Subject Interface
class Service(ABC):
    @abstractmethod
    def request(self):
        pass

# Real Subject
class RealService(Service):
    def request(self):
        return "RealService: Handling request"

# Proxy
class ProxyService(Service):
    def __init__(self):
        self._real_service = None

    def request(self):
        if self._real_service is None:
            print("Proxy: Initializing RealService...")
            self._real_service = RealService()
        print("Proxy: Logging access before forwarding request")
        return self._real_service.request()

# Usage
proxy = ProxyService()
print(proxy.request())
print(proxy.request())  # Second call uses cached RealService
```

## Key Takeaways
- The **Proxy pattern** provides a surrogate or placeholder for another object.
- It controls access to the real object (e.g., lazy initialization, logging, caching, security).
- Useful when direct access to the object is expensive or needs monitoring.
- Example: Proxy initializes `RealService` only once and logs every access.
- **Abstract Base Class (ABC)** is used to define an interface for `Service` and ensure subclasses implement `request()`.
