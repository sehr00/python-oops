# Day 10: Method Overloading in Python

In this notebook, we will learn:
- What method overloading means in Object‑Oriented Programming.
- Why Python does not support traditional overloading like Java or C++.
- How to achieve similar behavior using default arguments and *args.


```python
# Example: Method Overloading (Pythonic way)

class Calculator:
    def add(self, a=0, b=0, c=0):
        return a + b + c

# Usage
calc = Calculator()

print(calc.add(5, 10))        # adds two numbers
print(calc.add(5, 10, 15))    # adds three numbers
print(calc.add())             # adds nothing, returns 0
```

## Key Takeaways
- Python does not support traditional method overloading (same method name with different signatures).
- Instead, we use default arguments or *args to handle multiple cases.
- This makes methods flexible and able to accept varying numbers of parameters.
