# Day 2: Constructors and Methods in Python

In this notebook, we will deepen our understanding of Object-Oriented Programming (OOP) by focusing on:
- Constructors (`__init__`)
- Instance methods
- The `self` keyword


```python
# Example: Using constructors and methods

class Car:
    def __init__(self, brand, model, year):
        # Constructor initializes attributes
        self.brand = brand
        self.model = model
        self.year = year

    def description(self):
        # Instance method returns a formatted string
        return f"{self.year} {self.brand} {self.model}"

    def start_engine(self):
        return f"The {self.model}'s engine has started!"

# Creating objects
car1 = Car("Toyota", "Corolla", 2020)
car2 = Car("Tesla", "Model 3", 2022)

# Using methods
print(car1.description())
print(car2.start_engine())
```

## Key Concepts Learned
- **Constructor (`__init__`)**: Special method that runs automatically when an object is created.
- **Instance Method**: Functions defined inside a class that operate on object data.
- **self**: Refers to the current object instance.
