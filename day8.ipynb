# Day 8: Abstraction in Python

In this notebook, we will learn:
- What abstraction means in Object‑Oriented Programming.
- How to hide implementation details and show only essential features.
- How to use the `abc` module in Python to create abstract classes.


```python
# Example: Abstraction using abc module

from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        return "Car started with a key."

    def stop(self):
        return "Car stopped."

class Bike(Vehicle):
    def start(self):
        return "Bike started with a kick."

    def stop(self):
        return "Bike stopped."

# Usage
vehicles = [Car(), Bike()]

for v in vehicles:
    print(v.start())
    print(v.stop())
```

## Key Takeaways
- Abstraction hides implementation details and exposes only necessary methods.
- Abstract classes cannot be instantiated directly.
- Methods marked with `@abstractmethod` must be implemented in child classes.
- This ensures consistency across different subclasses.
