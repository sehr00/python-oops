# Day 7: Polymorphism in Python

In this notebook, we will learn:
- What polymorphism means in Object‑Oriented Programming.
- How the same method name can behave differently across classes.
- Why polymorphism makes code flexible and reusable.


```python
# Example: Polymorphism

class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

class Cow:
    def speak(self):
        return "Moo!"

# Using polymorphism
animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.speak())
```

## Key Takeaways
- Polymorphism allows different classes to define the same method name with different behavior.
- In the example, `speak()` behaves differently depending on whether the object is a Dog, Cat, or Cow.
- This makes code more flexible and easier to extend.
