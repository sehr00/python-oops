## What is Singleton Pattern?

The Singleton design pattern is a creational pattern that restricts the instantiation of a class to a single instance. This is useful when exactly one object is needed to coordinate actions across the system.

## Why use Singleton?

Singleton is used when we want to restrict the instantiation of a class to a single object. This is useful when exactly one object is needed to coordinate actions across the system.

## How to implement Singleton in Python?

There are several ways to implement Singleton in Python, but one common way is to use a class variable to store the single instance of the class.



```python
# Singleton using __new__ method

class Singleton:
    _instance = None  # class-level variable

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

# Usage
obj1 = Singleton()
obj2 = Singleton()

print("Object 1:", id(obj1))
print("Object 2:", id(obj2))
print("Same instance?", obj1 is obj2)
```

```python
#Alternative Singleton (Decorator)
def singleton(cls):
    instances = {}
    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]
    return get_instance

@singleton
class DatabaseConnection:
    def __init__(self):
        print("Connecting to database...")

# Usage
db1 = DatabaseConnection()
db2 = DatabaseConnection()
print("Same instance?", db1 is db2)
```

## Key Takeaways
- The **Singleton pattern** ensures only one instance of a class exists.
- Useful for shared resources (e.g., database connections, configuration managers).
- In Python, Singleton can be implemented using:
  - `__new__` method
  - Decorators
  - Metaclasses
- Singleton prevents duplication and ensures consistency across the application.
