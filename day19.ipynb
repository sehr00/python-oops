## What are Mixins?

Mixins are a type of class in Python that provides reusable functionality that can be added to other classes. They are similar to abstract classes in that they define methods that can be used by other classes, but they are not meant to be instantiated on their own.
Instead, they are used as a way to add functionality to other classes without having to inherit from them.

Mixins are useful for adding reusable functionality to multiple classes without having to duplicate code. For example, if you have a class that needs to handle user authentication, you could create a mixin class that provides the necessary methods for authentication, and then add that mixin to any class that needs to handle user authentication.

## How to create a Mixins
To create a mixin in Python, you can define a class that contains the methods you want to add to other classes. 



```python
# Example: LoggerMixin adds logging functionality

class LoggerMixin:
    def log(self, msg):
        print(f"[LOG]: {msg}")

class Service(LoggerMixin):
    def run(self):
        self.log("Service started")
        return "Service is running..."

class Payment(LoggerMixin):
    def process(self):
        self.log("Payment processed")
        return "Payment successful"

# Usage
s = Service()
p = Payment()

print(s.run())
print(p.process())
```

## Key Takeaways
- **Mixins** are small classes that provide reusable functionality.
- They are not meant to stand alone — instead, they are combined with other classes.
- Example: `LoggerMixin` adds logging ability to any class.
- Use Mixins when you want to share behavior across unrelated classes without deep inheritance hierarchies.
- Mixins can be added to classes using multiple inheritance.
