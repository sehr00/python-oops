In this notebook, we will learn:
- What aggregation means in OOP.
- How it differs from composition.
- Why aggregation allows objects to exist independently.
- How to implement aggregation in Python.

## What is aggregation?

Aggregation is a type of association between two classes. It is a "has-a" relationship. In aggregation, a class contains an object of another class. The object can exist independently of the container object.

For example, a car has an engine. The engine can exist independently of the car. The car can be sold without the engine, and the engine can be used in another car.

## How does aggregation differ from composition?       

Composition is a type of aggregation where the contained object cannot exist independently of the container object. For example, a car has an engine. The engine cannot exist independently of the car. The car cannot be sold without the engine.

## Why is aggregation useful?          

Aggregation allows objects to exist independently. This is useful in situations where an object may be used in multiple contexts. For example, an engine can be used in multiple cars. Aggregation also allows for more flexible code. For example, a car can be sold without an engine, and the engine can be added later.



```python
# Example: Aggregation (Employee belongs to a Department)

class Department:
    def __init__(self, name):
        self.name = name

class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department   # Aggregation: Employee HAS a Department

    def show(self):
        return f"{self.name} works in {self.department.name}"

# Usage
dept1 = Department("Computer Science")
dept2 = Department("Mathematics")

emp1 = Employee("misba", dept1)
emp2 = Employee("Midhat", dept2)

print(emp1.show())
print(emp2.show())
```

## Key Takeaways
- **Aggregation** is a Has‑A relationship where the contained object can exist independently.
- Example: An Employee *has a* Department, but the Department exists even without the Employee.
- Difference from composition:
  - Composition: contained object’s lifecycle is tied to the parent.
  - Aggregation: contained object can live on its own.
