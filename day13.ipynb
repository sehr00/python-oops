# Day 13: Class Methods and Static Methods

In this notebook, we will learn:
- The difference between instance methods, class methods, and static methods.
- How to use `@classmethod` and `@staticmethod` decorators.
- When to use each type of method in OOP design.


```python
# Example: Class Methods and Static Methods

class Student:
    school_name = "ABC High School"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance method (works with object data)
    def show(self):
        return f"{self.name} is {self.age} years old."

    # Class method (works with class-level data)
    @classmethod
    def change_school(cls, new_name):
        cls.school_name = new_name

    # Static method (utility function, no self/cls)
    @staticmethod
    def is_adult(age):
        return age >= 18

# Usage
s1 = Student("midhat", 17)
s2 = Student("mushtaq", 20)

print(s1.show())                  # Instance method
print(Student.school_name)         # Class variable

Student.change_school("XYZ Academy")
print(Student.school_name)         # Changed by class method

print(Student.is_adult(17))        # False
print(Student.is_adult(20))        # True
```

## Key Takeaways
- **Instance methods** use `self` and work with object data.
- **Class methods** use `cls` and work with class-level data. Declared with `@classmethod`.
- **Static methods** don’t use `self` or `cls`. Declared with `@staticmethod`. They are utility functions inside a class.
- Use class methods for factory patterns or modifying class state, and static methods for helper logic.
