## Project Overview

The Habit Tracker application will allow users to create and track their habits. Users can create habits, set reminders, and track their progress. The application will also provide analytics and insights to help users stay motivated and achieve their goals.

## Requirements

- Python 3.8 or higher

```python
# Final Project: Habit Tracker

# Singleton for Database
class Database:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.data = {}
            print("Database initialized.")
        return cls._instance

# Strategy for Reminder Methods
class ReminderStrategy:
    def remind(self, habit): pass

class EmailReminder(ReminderStrategy):
    def remind(self, habit):
        return f"Email reminder: Don't forget {habit}!"

class SMSReminder(ReminderStrategy):
    def remind(self, habit):
        return f"SMS reminder: Stay consistent with {habit}!"

# Observer for Notifications
class Observer:
    def update(self, message): pass

class User(Observer):
    def __init__(self, name):
        self.name = name
    def update(self, message):
        print(f"{self.name} received: {message}")

# Habit Tracker System
class HabitTracker:
    def __init__(self):
        self.habits = []
        self.users = []

    def add_habit(self, habit):
        self.habits.append(habit)
        self.notify(f"New habit added: {habit}")

    def attach(self, user):
        self.users.append(user)

    def notify(self, message):
        for u in self.users:
            u.update(message)

    def remind_all(self, strategy: ReminderStrategy):
        for habit in self.habits:
            print(strategy.remind(habit))

# Usage
db = Database()

tracker = HabitTracker()
user1 = User("Alice")
user2 = User("Bob")

tracker.attach(user1)
tracker.attach(user2)

tracker.add_habit("Morning Prayer")
tracker.add_habit("Daily Coding Practice")

tracker.remind_all(EmailReminder())
tracker.remind_all(SMSReminder())
```
