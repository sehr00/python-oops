## What is the Factory Method Pattern?

The Factory Method pattern is a creational design pattern that provides an interface for creating objects in a superclass, but allows subclasses to alter the type of objects that will be created. The Factory Method pattern is used to create objects without specifying the exact class of object that will be created.

## Why use the Factory Method Pattern?
The Factory Method pattern is used when a class cannot anticipate the class of objects it needs to create. It is also used when a class wants its subclasses to specify the objects it creates.

## How to implement the Factory Method Pattern in Python?
The Factory Method pattern is implemented by defining an interface for creating objects, and then implementing the interface in subclasses. The subclasses can then override the method to create objects of their own type.

## How to use the Factory Method Pattern in Python?
The Factory Method pattern is used by calling the factory method on an instance of the superclass. The factory method will then create an object of the appropriate type and return it. 


```python
# Factory Method Example

from abc import ABC, abstractmethod

# Abstract Product
class Document(ABC):
    @abstractmethod
    def create(self):
        pass

# Concrete Products
class PDFDocument(Document):
    def create(self):
        return "PDF Document created"

class WordDocument(Document):
    def create(self):
        return "Word Document created"

# Factory
class DocumentFactory(ABC):
    @abstractmethod
    def get_document(self):
        pass

class PDFDocumentFactory(DocumentFactory):
    def get_document(self):
        return PDFDocument()

class WordDocumentFactory(DocumentFactory):
    def get_document(self):
        return WordDocument()

# Usage
pdf_factory = PDFDocumentFactory()
word_factory = WordDocumentFactory()

pdf_doc = pdf_factory.get_document()
word_doc = word_factory.get_document()

print(pdf_doc.create())
print(word_doc.create())
```

## Key Takeaways
- The **Factory Method pattern** defines an interface for creating objects, but lets subclasses decide which class to instantiate.
- It promotes loose coupling by delegating object creation to factories.
- Useful when you want to centralize and standardize object creation logic.
- Example: `PDFDocumentFactory` and `WordDocumentFactory` create different types of documents without changing client code.
- **Abstract Factory** pattern is an extension of Factory Method, providing a way to produce families of related objects without specifying their concrete classes.
- **Builder** pattern focuses on constructing complex objects step by step, allowing for flexibility and customization.
- **Prototype** pattern allows for the creation of new objects by copying existing ones, which can be more efficient than creating from scratch.
- **Singleton** pattern ensures a class has only one instance and provides a global point of access to it.
