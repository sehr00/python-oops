## Namespaces

A namespace is a container for a set of objects, such as variables and functions, that are defined in a particular scope. In Python, namespaces are used to organize code and prevent name collisions.



```python
# Example: Global vs Local scope

x = 10  # Global variable

def outer_function():
    x = 20  # Enclosing variable

    def inner_function():
        x = 30  # Local variable
        print("Inner:", x)  # Local scope

    inner_function()
    print("Outer:", x)  # Enclosing scope

outer_function()
print("Global:", x)  # Global scope
```

```python
# Built-in scope example
print(len([1, 2, 3]))  # len is a built-in function
```

## Key Takeaways
- Python resolves variables using the **LEGB rule**:
  - Local → Enclosing → Global → Built-in.
- Local scope: inside a function.
- Enclosing scope: outer function variables.
- Global scope: defined at the top level of a module.
- Built-in scope: Python’s built-in names (e.g., len, print).
- Understanding scope prevents variable conflicts and bugs.
- Namespaces help organize code and prevent name collisions.
