## What is Adapter Pattern?
The Adapter design pattern is a structural design pattern that allows objects with incompatible interfaces to work together. It is used to convert the interface of a class into another interface that clients expect. This pattern is useful when you have a class that you cannot modify or when you want to use a class that has an incompatible interface.

## Why use Adapter Pattern?
The Adapter design pattern is useful in the following scenarios:
- When you have a class that you cannot modify or when you want to use a class that has an incompatible interface.
- When you want to reuse existing classes and extend their functionality without modifying their code.
- When you want to create a flexible and extensible system that can work with different classes with different interfaces.

## How to implement Adapter Pattern in Python?
To implement the Adapter pattern in Python, you can follow these steps:
1. Define the target interface that the client expects to use.
2. Define the adaptee class that has the incompatible interface.
3. Define the adapter class that implements the target interface and delegates the calls to the adaptee class.
4. Use the adapter class in the client code to access the adaptee class through the target interface.


```python
# Adapter Pattern Example

# Existing class (incompatible interface)
class OldPrinter:
    def print_text(self, text):
        return f"Old Printer: {text}"

# Desired interface
class NewPrinterInterface:
    def print(self, message):
        pass

# Adapter
class PrinterAdapter(NewPrinterInterface):
    def __init__(self, old_printer: OldPrinter):
        self.old_printer = old_printer

    def print(self, message):
        # Translate the call to the old interface
        return self.old_printer.print_text(message)

# Usage
old_printer = OldPrinter()
adapter = PrinterAdapter(old_printer)

print(adapter.print("Hello, Adapter Pattern!"))
```

## Key Takeaways
- The **Adapter pattern** allows incompatible interfaces to work together.
- It acts as a translator between the client and the existing class.
- Useful when integrating legacy code with new systems.
- Example: An adapter lets a new interface call methods of an old printer class.
