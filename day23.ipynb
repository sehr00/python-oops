## What is the Observer Pattern?
The Observer design pattern is a behavioral design pattern that allows an object, called the subject, to notify other objects, called observers, when its state changes. The observers can then take appropriate actions based on the change in the subject's state.

The Observer pattern is useful when you want to decouple the subject from its observers. This means that the subject can notify the observers without knowing anything about them, and the observers can subscribe to the subject without knowing anything about it.

## Why use the Observer Pattern?
The Observer pattern is useful in situations where you want to create a one-to-many relationship between objects. For example, you might have a subject that represents a data source, and multiple observers that need to be notified when the data changes. By using the Observer pattern, you can decouple the data source from its observers, making it easier to add or remove observers without affecting the data source.

## How to Implement the Observer Pattern in Python?
To implement the Observer pattern in Python, you can use the following steps:
1. Define the Observer interface or base class.
2. Define the Subject interface or base class.

```python
# Observer Pattern Example

# Subject (Publisher)
class Subject:
    def __init__(self):
        self._observers = []

    def attach(self, observer):
        self._observers.append(observer)

    def detach(self, observer):
        self._observers.remove(observer)

    def notify(self, message):
        for observer in self._observers:
            observer.update(message)

# Observer (Subscriber)
class Observer:
    def __init__(self, name):
        self.name = name

    def update(self, message):
        print(f"{self.name} received: {message}")

# Usage
subject = Subject()

observer1 = Observer("Observer 1")
observer2 = Observer("Observer 2")

subject.attach(observer1)
subject.attach(observer2)

subject.notify("Dhikr reminder: SubhanAllah")
```

## Key Takeaways
- The **Observer pattern** defines a one-to-many dependency between objects.
- When the subject changes state, all observers are notified automatically.
- Useful for event-driven systems, GUIs, and notification services.
- Example: A Dhikr reminder system notifies multiple users at once.
- **Subject** class maintains a list of observers and notifies them.
- **Observer** class defines an update method to handle notifications.
- **attach()** and **detach()** methods manage observer subscriptions.
- **notify()** method sends messages to all subscribed observers.
