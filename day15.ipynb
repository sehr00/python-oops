# Day 15: Composition (Has‑A Relationship)

In this notebook, we will learn:
- What composition means in OOP.
- How one class can contain another class as part of its attributes.
- Why composition is different from inheritance.


```python
# Example: Composition (Car has an Engine)

class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self, model):
        self.model = model
        self.engine = Engine()   # Composition: Car HAS an Engine

    def drive(self):
        return f"{self.model}: {self.engine.start()} → Car is moving"

# Usage
c1 = Car("Toyota Corolla")
c2 = Car("Honda Civic")

print(c1.drive())
print(c2.drive())
```

```python
## Key Takeaways
- **Composition** means one class contains another class (Has‑A relationship).
- Example: A Car *has an* Engine.
- It allows building complex objects by combining simpler ones.
- Use composition when you want to reuse functionality without inheritance.
```
