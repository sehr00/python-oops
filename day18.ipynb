## What is an interface?

An interface is a contract that a class agrees to implement. It defines a set of methods that a class must implement. Any class that implements an interface is said to conform to that interface.

## Why do we need interfaces?
Interfaces are useful for:
- Ensuring that a class has a certain set of methods.
- Providing a way to define a common interface for a group of classes.
- Facilitating polymorphism.

## How do we define an interface in Python?

In Python, we can define an interface using Abstract Base Classes (ABCs). An ABC is a class that cannot be instantiated and is used as a base class for other classes. It defines a set of abstract methods that must be implemented by any class that inherits from it.

## How do we use interfaces in Python?
To use an interface in Python, we can create an ABC and define the abstract methods that must be implemented by any class that inherits from it. We can then create a class that inherits from the ABC and implements the abstract methods.




```python
from abc import ABC, abstractmethod

# Define an interface using ABC
class Printable(ABC):
    @abstractmethod
    def print(self):
        pass

# Concrete class implementing the interface
class Report(Printable):
    def print(self):
        return "Printing report..."

class Invoice(Printable):
    def print(self):
        return "Printing invoice..."

# Usage
docs = [Report(), Invoice()]

for d in docs:
    print(d.print())
```

```python
# Duck typing: no formal interface, just method presence
class PDF:
    def print(self):
        return "PDF document printed"

class WordDoc:
    def print(self):
        return "Word document printed"

def print_document(doc):
    print(doc.print())

# Usage
print_document(PDF())
print_document(WordDoc())
```

## Key Takeaways
- Python doesn’t have formal interfaces like Java or C#.
- You can mimic interfaces using Abstract Base Classes (ABCs).
- Duck typing allows objects to be used as long as they implement the required methods.
- Interfaces enforce consistency across different classes.
- Duck typing allows for more flexibility and less strict adherence to a specific structure.
